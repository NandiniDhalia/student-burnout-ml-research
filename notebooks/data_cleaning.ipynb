# Encoding the categorical features

```python
df_encoded=df.copy()
categorical_cols=['gender','year','burnout_level','sleep_quality']
encoder=LabelEncoder()
for col in categorical_cols:
  df_encoded[col]=encoder.fit_transform(df_encoded[col])
```

```python
df_encoded.select_dtypes(include='object').columns
```

```python
df_encoded['stress_level'].unique()
```

```python
stress_map={'High':2,'Medium':1,'Low':0}
df_encoded['stress_level']=df_encoded['stress_level'].map(stress_map)
```

# Correlation Check

# Scaling

```python
numerical_cols=['age','daily_study_hours','daily_sleep_hours','screen_time_hours','stress_level','anxiety_score','academic_pressure_score','social_support_score','physical_activity_hours']
scaler=StandardScaler()
df_encoded[numerical_cols]=scaler.fit_transform(df_encoded[numerical_cols])
```
